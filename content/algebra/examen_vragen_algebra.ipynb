### A1. Vectoroperaties

Gegeven `u = np.array([2, -1, 4, 3])` en `v = np.array([1, 5, -2, 0])`.

Bereken het **dot product** van `u` en `v`, en bereken de **hoek** (in graden) tussen beide vectoren.

````python
u = np.array([2, -1, 4, 3])
v = np.array([1, 5, -2, 0])

# Jouw code:
# Dot product berekenen:
dot_num = np.dot(u, v)
print(f"Dot product: {dot_num}")

# Hoek berekenen:
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

# cos(θ) = (u · v) / (||u|| × ||v||)
cos_theta = dot_num / (norm_u * norm_v)

# θ = arccos(cos(θ))
theta_radialen = np.arccos(cos_theta)
theta_graden = np.degrees(theta_radialen)
print(f"Hoek: {theta_graden}°")
````

### A2. Norm en normalisatie

Normaliseer de vector `w = np.array([3, 0, 4, 0])` zodat deze lengte 1 heeft. Print de genormaliseerde vector en **bewijs** dat de norm inderdaad 1 is.

````python
w = np.array([3, 0, 4, 0])

# Jouw code:
# Norm van w
norm_w = np.linalg.norm(w)
print(f"Norm van w: {norm_w}")

# Genormaliseerde vector
w_genormaliseerd = w / norm_w
print(f"Genormaliseerded vector: {w_genormaliseerd}")

# Norm van genormaliseerde vector
print(f"Norm van genormaliseerde vector: {np.linalg.norm(w_genormaliseerd):.10f}")

# Lengte
print("De genormaliseerde vector heeft exact lengte 1.")
````

### A4. Determinant en inverteerbaarheid

Bereken de determinant van de volgende twee matrices. Welke is inverteerbaar en waarom?

```
M1 = [[6, 3],      M2 = [[4, 2],
      [4, 2]]            [7, 5]]
```

````python
M1 = np.array([[6, 3], [4, 2]])
M2 = np.array([[4, 2], [7, 5]])

# Jouw code:
# Determinant M1
det_num = np.linalg.det(M1)

# Determinant M2
det_num2 = np.linalg.det(M2)

# Print()
print(f"Determinant M1: {det_num}")
print(f"Determinant M2: {det_num2}")
print("M2 is inverteerbaar omdat de determinant geen 0.0 is zoals M1.")
````

### A5. Stelsel oplossen

Los het volgende stelsel op met `np.linalg.solve()` en verifieer je antwoord:

```
3x + 2y - z = 1
 x -  y + 2z = 5
2x + 3y +  z = 8
```

````python
# Jouw code:
# Het stelsel in matrixnotatie
A = np.array([[3, 2, -1],
              [1, -1, 2],
              [2, 3, 1]])

b = np.array([1, 5, 8])

# Oplossen met NumPy
x_oplossing = np.linalg.solve(A, b)

print(f"Oplossing: x = {x_oplossing}")
print(f"  x = {x_oplossing[0]}")
print(f"  y = {x_oplossing[1]}")
print(f"  z = {x_oplossing[2]}")
print()

# Verificatie
print("Verificatie (A @ x moet gelijk zijn aan b):")
print(f"  A @ x = {A @ x_oplossing}")
print(f"  b     = {b}")
print(f"  Gelijk? {np.allclose(A @ x_oplossing, b)}")
````

### A6. Inverse berekenen en verifiëren

Bereken de inverse van de matrix hieronder en toon aan dat `A @ A_inv` gelijk is aan de identiteitsmatrix.

```
A = [[2, 1],
     [5, 3]]
```

````python
A = np.array([[2, 1], [5, 3]])

# Jouw code:
A_inv = np.linalg.inv(A)

print("Matrix A:")
print(A)
print()
print("Inverse A⁻¹:")
print(A_inv)
print()

# Verificatie: A × A⁻¹ = I
print("Verificatie: A @ A⁻¹ =")
print(A @ A_inv)
print()
print("Verificatie: A⁻¹ @ A =")
print(A_inv @ A)
print()
print(f"Beide zijn de identiteitsmatrix? {np.allclose(A @ A_inv, np.eye(2)) and np.allclose(A_inv @ A, np.eye(2))}")
````

### A8. Neuron output

Een neuron heeft gewichten `w = [0.5, -0.3, 0.8]`, bias `b = 0.2`, en ontvangt input `x = [1.0, 2.0, 0.5]`.

Bereken de output van dit neuron **met** ReLU activatie.

````python
w = np.array([0.5, -0.3, 0.8])
b = 0.2
x = np.array([1.0, 2.0, 0.5])

# Jouw code:
z = np.dot(w, x) + b
print(f"Lineaire outpur z: {z}")

output = np.maximum(0, z)
print(f"Neuron output (ReLu): {output}")
````

### A10. Determinant eigenschap

Gegeven een willekeurige 3×3 matrix `A` en scalar `c = 3`, verifieer dat:

**det(c · A) = c³ · det(A)**

Leg in een commentaar uit waarom de exponent 3 is.

````python
# Jouw code:
A = np.random.randn(3, 3)
c = 3

left = np.linalg.det(c * A)
right = (c**3) * np.linalg.det(A)

# Print()
print(f"det(c * A): {left}")
print(f"c³ * det(A): {right}")
print(f"Eigenschap klopt: {np.isclose(left, right)}")

# De exponent is 3 omdat mijn matrix is 3x3.
# Als je een matrix vermenigvuldigt met een scalar, dan wordt elke rij vermenigvuldigd met c.
````
